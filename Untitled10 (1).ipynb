```python
!pip install -q catboost

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier, XGBRFClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PowerTransformer
from sklearn.compose import ColumnTransformer

# Correcting categorical columns to match the dataset (removed 'Name' as it has been dropped)
categorical_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'Side', 'Cabin_num', 'VIP']
u_cols = ['Age',  'RoomService', 'FoodCourt', 'ShoppingMall' ,'Spa', 'VRDeck']

categorical_pipeline = Pipeline(steps=[
    ('imputes', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

numerical_pipeline = Pipeline(steps=[
    ('imputes', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('power', PowerTransformer())
])

preprocessor = ColumnTransformer(transformers=[
    ('categorical', categorical_pipeline, categorical_cols),
    ('numerical', numerical_pipeline, u_cols)
])

models = {
    "Logistic Regression" : LogisticRegression( random_state=42),
    "Decision Tree"       : DecisionTreeClassifier(random_state=42),
    "Random Forest"       : RandomForestClassifier(random_state=42),
    "Gradient Boosting"   : GradientBoostingClassifier(random_state=42),
    "XGBoost"             : XGBClassifier(random_state=42, eval_metric='logloss'),
    "LightGBM"            : LGBMClassifier(random_state=42, verbose=-1),
    "CatBoost"            : CatBoostClassifier(random_state=42, verbose=0)
}

for name, model in models.items():
    pipe = Pipeline(steps=[("preprocessor", preprocessor), ("classifier", model)])

    accuracy = cross_val_score(pipe, x_train, y_train, cv=3, scoring="accuracy")
    macrof1 = cross_val_score(pipe, x_train, y_train, cv=3, scoring="f1_macro")

    print(f"{name} -> Accuracy: {accuracy.mean():.3f} , F1: {macrof1.mean():.3f}")
```

```python

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PowerTransformer
from sklearn.compose import ColumnTransformer

categorical_cols = [
    'HomePlanet', 'CryoSleep', 'Destination',
    'Side', 'Cabin_num', 'VIP'
]

numerical_cols = [
    'Age', 'RoomService', 'FoodCourt',
    'ShoppingMall', 'Spa', 'VRDeck'
]

# Check columns before training
missing_cat = set(categorical_cols) - set(x_train.columns)
missing_num = set(numerical_cols) - set(x_train.columns)

if missing_cat or missing_num:
    raise ValueError(
        f"Missing categorical columns: {missing_cat}\n"
        f"Missing numerical columns: {missing_num}\n"
        "Check your feature engineering and column names."
    )

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('power', PowerTransformer())
])

preprocessor = ColumnTransformer([
    ('categorical', categorical_pipeline, categorical_cols),
    ('numerical', numerical_pipeline, numerical_cols)
])

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),
    "XGBoost": XGBClassifier(
        random_state=42, eval_metric='logloss'
    ),
    "LightGBM": LGBMClassifier(
        random_state=42, verbosity=-1
    ),
    "CatBoost": CatBoostClassifier(
        random_state=42, verbose=0
    )
}

cv = StratifiedKFold(
    n_splits=3, shuffle=True, random_state=42
)

for name, model in models.items():
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    scores = cross_validate(
        pipe,
        x_train,
        y_train,
        cv=cv,
        scoring=['accuracy', 'f1_macro'],
        n_jobs=1,
        error_score='raise'
    )

    print(
        f"{name} -> "
        f"Accuracy: {scores['test_accuracy'].mean():.3f} "
        f"+/- {scores['test_accuracy'].std():.3f}, "
        f"Macro F1: {scores['test_f1_macro'].mean():.3f}"
    )
```
